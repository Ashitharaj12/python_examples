*Upload image using TorchVision*

```python
import torch
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt

#load an image into IDE
image_path ="/content/classroom.jpg"
image = Image.open(image_path)

image=Image.open(image_path)

#define transformation
transform=transforms.Compose([
    transforms.Resize((128,128)),
    transforms.ToTensor(), #convert to tenser
])

#apply transformation
image_tensor=transform(image)

#Convert tensor back to image for display as an output
image_numpy=image_tensor.permute(1,2,0).numpy()

#display the image
plt.imshow(image_numpy)
plt.axis('off')
plt.title("transformed images")
plt.show()
```
