```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from google.colab import files
import pandas as pd
uploaded = files.upload()
```

```python
df=pd.read_csv('crop_yield_data.xls')
X = df.drop('crop_yield', axis=1)  # Features
y = df['crop_yield']  # Target variable
```

```python
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
model = Sequential()
model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))  # For regression output
```

```python
model.compile(optimizer='adam', loss='mean_squared_error')
```

```python
model.fit(X_train, y_train, epochs=100, batch_size=10)
```

```python
loss = model.evaluate(X_test, y_test)
print(f"Test Loss: {loss}")
```

```python
#example program on images using MNIST dataset
import matplotlib.pyplot as plt
from tensorflow import keras

#load MNIST dataset (predefined)
(x_train, y_train),(x_test,y_test)=keras.datasets.mnist.load_data()

#display functions for images
def show_images(images,labels, num=10):
    plt.figure(figsize=(10,5))
    for i in range(num):
      plt.subplot(2, 5, i+1) #print 10 images in 2 different lines
      plt.imshow(images[i], cmap='jet')
      plt.title(f"label :{labels[i]}")
      plt.axis('off')
      plt.show()
show_images(x_train, y_train)
```
